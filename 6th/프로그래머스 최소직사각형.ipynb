```python
import ast
def solution(sizes):
    n = len(sizes)
    maxw = 0
    maxh = 0
    for i in range (n):
        w = sizes[i][0] #가로
        h = sizes[i][1] #세로
        if w < h :
            z = 0
            z = w
            w = h  #더 긴 것이 가로로 회전
            h = z
        if w > maxw :  #max x보다 크면 바꿈
            maxw = w
        if h > maxh :  #max y보다 크면 바꿈
            maxh = h
    answer = maxw * maxh
    return answer
```
