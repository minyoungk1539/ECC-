```python
def solution(edges):
    graph = {}
    indeg = {}
    outdeg = {}

    for a, b in edges:
        if a not in graph:
            graph[a] = []
        graph[a].append(b)

        outdeg[a] = outdeg.get(a, 0) + 1
        indeg[b] = indeg.get(b, 0) + 1

        if b not in outdeg:
            outdeg[b] = outdeg.get(b, 0)
        if a not in indeg:
            indeg[a] = indeg.get(a, 0)

    created = 0
    donut = 0
    stick = 0
    eight = 0

    for node in outdeg:
        if indeg[node] == 0 and outdeg[node] >= 2:
            created = node
            break

    for node in outdeg:
        if node == created:
            continue

        if outdeg[node] == 0:
            stick += 1
        elif outdeg[node] == 2:
            eight += 1

    total = outdeg[created]
    donut = total - stick - eight

    return [created, donut, stick, eight]
```
